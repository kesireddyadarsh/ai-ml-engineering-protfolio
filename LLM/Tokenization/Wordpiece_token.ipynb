```python
word_splits = {}

for word in word_frequency:
    pieces = []

    for index, character in enumerate(word):
        if index == 0:
            pieces.append(character)
        else:
            pieces.append("##" + character)

    word_splits[word] = pieces

print(word_splits["machine"])
print(word_splits["learning"])
```

```python
pair_frequency = {}

for word, pieces in word_splits.items():
    frequency = word_frequency[word]

    for index in range(len(pieces) - 1):
        pair = (pieces[index], pieces[index + 1])
        pair_frequency[pair] = pair_frequency.get(pair, 0) + frequency

sorted_pairs = sorted(
    pair_frequency.items(),
    key=lambda item: item[1],
    reverse=True
)

print(sorted_pairs[:20])
```

```python
token_frequency = {}

for word, pieces in word_splits.items():
    frequency = word_frequency[word]

    for token in pieces:
        token_frequency[token] = token_frequency.get(token, 0) + frequency
```

```python
pair_scores = {}

for pair, pair_count in pair_frequency.items():
    first_token, second_token = pair

    score = pair_count / (
        token_frequency[first_token] *
        token_frequency[second_token]
    )

    pair_scores[pair] = score
```

```python
sorted_scores = sorted(
    pair_scores.items(),
    key=lambda item: item[1],
    reverse=True
)

print(sorted_scores[:20])
```

```python
best_pair = max(pair_scores, key=pair_scores.get)

print("Best pair:", best_pair)
print("Score:", pair_scores[best_pair])
```

```python
initial_vocabulary.add(merged_token)

print("Vocabulary size:", len(initial_vocabulary))
```

```python
target_vocab_size = 500

while len(initial_vocabulary) < target_vocab_size:

    token_frequency = {}
    pair_frequency = {}

    # Recalculate token and pair frequencies
    for word, pieces in word_splits.items():
        frequency = word_frequency[word]

        for token in pieces:
            token_frequency[token] = (
                token_frequency.get(token, 0) + frequency
            )

        for index in range(len(pieces) - 1):
            pair = (pieces[index], pieces[index + 1])

            pair_frequency[pair] = (
                pair_frequency.get(pair, 0) + frequency
            )

    # Stop if no pairs remain
    if not pair_frequency:
        break

    # Calculate WordPiece pair scores
    pair_scores = {}

    for pair, pair_count in pair_frequency.items():
        first_token, second_token = pair

        score = pair_count / (
            token_frequency[first_token]
            * token_frequency[second_token]
        )

        pair_scores[pair] = score

    # Select highest-scoring pair
    best_pair = max(pair_scores, key=pair_scores.get)

    first_token, second_token = best_pair

    # Create merged token
    if second_token.startswith("##"):
        merged_token = first_token + second_token[2:]
    else:
        merged_token = first_token + second_token

    # Update all word splits
    for word, pieces in word_splits.items():
        new_pieces = []
        index = 0

        while index < len(pieces):
            if (
                index < len(pieces) - 1
                and pieces[index] == first_token
                and pieces[index + 1] == second_token
            ):
                new_pieces.append(merged_token)
                index += 2
            else:
                new_pieces.append(pieces[index])
                index += 1

        word_splits[word] = new_pieces

    initial_vocabulary.add(merged_token)

print("Final vocabulary size:", len(initial_vocabulary))
```

```python
special_tokens = ["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"]

final_vocabulary = special_tokens + sorted(initial_vocabulary)

with open("vocab.txt", "w", encoding="utf-8") as file:
    for token in final_vocabulary:
        file.write(token + "\n")

print("Saved vocabulary size:", len(final_vocabulary))
```

```python
with open("vocab.txt", "r", encoding="utf-8") as file:
    vocabulary = {line.strip() for line in file}

print("Loaded vocabulary size:", len(vocabulary))
```

```python
def wordpiece_tokenize_word(word, vocabulary):
    if word in vocabulary:
        return [word]

    tokens = []
    start = 0

    while start < len(word):
        end = len(word)
        matched_token = None

        while start < end:
            piece = word[start:end]

            if start > 0:
                piece = "##" + piece

            if piece in vocabulary:
                matched_token = piece
                break

            end -= 1

        if matched_token is None:
            return ["[UNK]"]

        tokens.append(matched_token)
        start = end

    return tokens
```

```python
print(wordpiece_tokenize_word("learning", vocabulary))
print(wordpiece_tokenize_word("tokenization", vocabulary))
print(wordpiece_tokenize_word("unknownword", vocabulary))
```

```python
custom_tokens = []

for word in user_input.lower().split():
    word_tokens = wordpiece_tokenize_word(word, vocabulary)
    custom_tokens.extend(word_tokens)

print(custom_tokens[:100])
print("Total tokens:", len(custom_tokens))
```
