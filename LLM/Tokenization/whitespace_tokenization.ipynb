```python
user_input = """
Machine learning represents a paradigm shift in software development where computer systems analyze vast quantities of data to recognize patterns and make decisions without being explicitly programmed with rigid rules. At its core, machine learning operates across three foundational paradigms depending on the nature of the data and the desired outcome. Supervised learning utilizes labeled datasets to map known inputs to target outputs, making it ideal for classification and regression tasks like predicting financial trends or detecting spam. Unsupervised learning analyzes unlabeled datasets to discover inherent groupings, hidden structures, or anomalies without human guidance, which is essential for tasks such as customer segmentation. Reinforcement learning relies on an agent interacting with a dynamic environment, learning optimal decision-making strategies through trial and error based on a system of rewards and penalties.
Deep learning is a specialized subset of machine learning that draws architectural inspiration from the structure and functioning of the biological human brain through multi-layered artificial neural networks. Unlike traditional machine learning techniques that frequently depend on human experts to manually engineer features from raw data, deep learning algorithms automatically learn hierarchical abstractions directly from high-dimensional datasets. By passing information through sequential layers of mathematical nodes, early layers capture simple attributes while deeper layers synthesize these into highly abstract concepts, enabling breakthroughs in complex domains such as computer vision, speech recognition, and medical diagnostics where traditional algorithmic approaches historically struggled.
Large language models represent a highly specialized and powerful subfield within deep learning, designed specifically to process, understand, and generate natural human language. These models are constructed using advanced neural network architectures known as transformers, which excel at modeling sequential data and capturing long-range dependencies between words through self-attention mechanisms. Trained on immense corpora containing billions or trillions of text tokens, large language models learn the underlying statistical relationships, grammar, context, and semantic nuances of language. This deep contextual comprehension allows them to perform a vast range of downstream tasks, from answering complex queries and translating languages to writing software code and facilitating natural conversational interaction.
Python serves as the primary programming language unifying this entire artificial intelligence hierarchy, acting as the universal industry standard across research, development, and deployment. Python's clear syntax and emphasis on code readability allow researchers and software engineers to focus on designing complex algorithm architectures rather than managing low-level hardware memory. Furthermore, Python boasts a massive and mature ecosystem of specialized libraries, including scikit-learn for classical machine learning as well as PyTorch and TensorFlow for deep learning and large language models. While Python itself is an interpreted language, these framework libraries execute heavy mathematical operations using underlying computational backends written in C, C++, and CUDA, effectively combining developer efficiency with high-performance GPU hardware acceleration.
"""
```

```python
tokens = user_input.split()
print(len(tokens))
```

```python
#Creating a vocabulary
vocabulary = {}

for text in tokens:
    vocabulary[text] = vocabulary.get(text, 0 ) + 1

print(vocabulary)
```

```python
token_ids = [vocabulary[token] for token in tokens]
print(token_ids)
```

```python
import re
user_input_lower = re.sub(r'[^a-zA-Z]', ' ',user_input.lower())
tokens_lower = user_input_lower.split()
vocabulary_lower = {}
for text in tokens_lower:
    vocabulary_lower[text] = vocabulary_lower.get(text, 0) + 1

print(vocabulary_lower)
```
