```python
with open("sample_text.txt", "r", encoding="utf-8") as file:
    sample_text = file.read()

print(sample_text[:500])
```

Word-frequency Dictionary

```python
from collections import Counter

words = sample_text.split()

word_freq = Counter(words)

print(word_freq)
```

```python
word_tokens = {}

for word in word_freq:
    word_tokens[word] = ["▁"] + list(word)
```

```python
for word, tokens in word_tokens.items():
    print(word, tokens, word_freq[word])
```

```python
from collections import defaultdict

pair_freq = defaultdict(int)

for word, tokens in word_tokens.items():
    freq = word_freq[word]

    for i in range(len(tokens) - 1):
        pair = (tokens[i], tokens[i + 1])
        pair_freq[pair] += freq

print(pair_freq)

best_pair = max(pair_freq, key=lambda pair: pair_freq[pair])

print("Best pair:", best_pair)
print("Frequency:", pair_freq[best_pair])
```

```python
def merge_pair(tokens, pair):
    new_tokens = []
    i = 0

    while i < len(tokens):
        if (
            i < len(tokens) - 1
            and tokens[i] == pair[0]
            and tokens[i + 1] == pair[1]
        ):
            new_tokens.append(tokens[i] + tokens[i + 1])
            i += 2
        else:
            new_tokens.append(tokens[i])
            i += 1

    return new_tokens

for word in word_tokens:
    word_tokens[word] = merge_pair(
        word_tokens[word],
        best_pair
    )
```

```python
vocab = set()

for tokens in word_tokens.values():
    vocab.update(tokens)

target_vocab_size = 100
merges = []

while len(vocab) < target_vocab_size:

    pair_freq = defaultdict(int)

    for word, tokens in word_tokens.items():
        freq = word_freq[word]

        for i in range(len(tokens) - 1):
            pair = (tokens[i], tokens[i + 1])
            pair_freq[pair] += freq

    if not pair_freq:
        break

    best_pair, best_frequency = max(
        pair_freq.items(),
        key=lambda item: item[1]
    )

    new_token = best_pair[0] + best_pair[1]

    merges.append(best_pair)
    vocab.add(new_token)

    for word in word_tokens:
        word_tokens[word] = merge_pair(
            word_tokens[word],
            best_pair
        )

    print(
        "Merged:",
        best_pair,
        "->",
        new_token,
        "Frequency:",
        best_frequency,
        "Vocab size:",
        len(vocab)
    )
```

```python
print("Vocabulary size:", len(vocab))

for token in sorted(vocab):
    print(token)
```

```python
vocab_dict = {
    token: idx
    for idx, token in enumerate(sorted(vocab))
}

print(vocab_dict)
```

```python
word = "machines"
tokens = list(word)
print(tokens)

for pair in merges:
    tokens = merge_pair(tokens, pair)

print(tokens)

token_ids = [vocab_dict[token] for token in tokens]

print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

```python
def bpe_tokenize(text):
    final_tokens = []

    words = text.split()

    for word in words:
        tokens = list(word)

        for pair in merges:
            tokens = merge_pair(tokens, pair)

        final_tokens.extend(tokens)

    return final_tokens

text = "machine learning is useful"
tokens = bpe_tokenize(text)
print("Tokens:", tokens)


token_ids = [vocab_dict[token] for token in tokens]
print("Token IDs:", token_ids)

```
