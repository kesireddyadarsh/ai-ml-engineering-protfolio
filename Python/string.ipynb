```python
import re
def first_non_repeat(string_find):
    return_dict = dict()
    for char in string_find:
        return_dict[char] = return_dict.get(char,0) + 1
    for key in return_dict:
        if return_dict[key] == 1:
            return key
    return None


string_find = "Swiss"
string_find = re.sub(r'[^a-zA-Z]', '',string_find.lower())
print(string_find)
first_non_repeat(string_find)
```

```python
import re

def anagrams(*args):
    lst_input = list(args)
    #check for constant length
    size_c = len(lst_input[0])
    for c in lst_input:
        if len(c) != size_c:
            return False
    
    #check for values 
    tot_dict = list()
    for temp in lst_input:
        temp_dict = dict()
        for char in temp:
            temp_dict[char] = temp_dict.get(char, 0) +1
        tot_dict.append(temp_dict)
    
    #check if each has same key
    reference = tot_dict[0]

    for dictionary in tot_dict[1:]:
        if reference != dictionary:
            return False

    return True
            

str_1 = "Dormitory"
str_2 = "Dirty Room"
str_1 = re.sub(r'[^a-zA-Z]','',str_1.lower())
str_2 = re.sub(r'[^a-zA-Z]','',str_2.lower())
anagrams(str_1,str_2)
```

```python
import re

def noDublicates(text):
    seen = set()
    result = []

    for char in text.lower():
        if char not in seen:
            seen.add(char)
            result.append(char)

    return ''.join(result)

str_v = "Hello, World!!"
str_v = re.sub(r'[^a-zA-Z]','',str_v.lower())
noDublicates(str_v)
```

```python
counts = {
    "vowels": 0,
    "consonants": 0,
    "digits": 0,
    "spaces": 0,
    "special_characters": 0
}

str_v = "Python3 is Awesome!!"
print(list(str_v.lower()))
vowels = 'aeiou'
for value in str_v:
    if value in vowels:
        counts["vowels"] += 1
    elif value.isspace():
        counts["spaces"] += 1
    elif value.isdigit():
        counts["digits"] += 1
    elif value.isalpha():
        counts["consonants"] += 1
    else:
        counts["special_characters"] += 1

print(counts)
```

```python
def snake_case(text):
    text= list(text)
    for index,value in enumerate(text):
        if value.isupper():
            text[index] = f'_'+value.lower()
    return "".join(text)
def camelcase(text):
    result = []
    capitalize_next = False

    for char in text:
        if char == "_":
            capitalize_next = True
            continue

        if capitalize_next:
            result.append(char.upper())
            capitalize_next = False
        else:
            result.append(char)

    return "".join(result)

snake_case("machineLearningEngineer")
camelcase('machine_learning_engineer')
```
