```python
students = [
    {"name": "Alice", "department": "CS"},
    {"name": "Bob", "department": "Math"},
    {"name": "Charlie", "department": "CS"},
    {"name": "David", "department": "Physics"},
    {"name": "Eva", "department": "Math"},
]

depart_dict = {}
for individual in students:
    if individual["department"] not in depart_dict:
        depart_dict[individual["department"]] = [individual["name"]]
    else:
        depart_dict[individual["department"]].append(individual["name"])

print(depart_dict)
```

```python
students = [
    {"name": "Alice", "department": "CS", "gpa": 3.8},
    {"name": "Bob", "department": "Math", "gpa": 3.2},
    {"name": "Charlie", "department": "CS", "gpa": 3.5},
    {"name": "David", "department": "Physics", "gpa": 3.9},
    {"name": "Eva", "department": "Math", "gpa": 3.6},
]
depart_dict = {}
for individual in students:
    if individual["department"] not in depart_dict:
        depart_dict[individual["department"]] = {
            'count': 1,
            'total_gpa': individual["gpa"],
            "highest_gpa": individual["gpa"]
        }
    else:
        depart_dict[individual["department"]]['count'] += 1
        depart_dict[individual["department"]]['total_gpa'] += individual["gpa"]
        if depart_dict[individual["department"]]['highest_gpa'] < individual["gpa"]:
            depart_dict[individual["department"]]['highest_gpa'] = individual["gpa"]

for department in depart_dict:
    depart_dict[department]["average_gpa"] = (
        depart_dict[department]["total_gpa"]
        / depart_dict[department]["count"]
    )

print(depart_dict)
```

```python
inventory = {
    "Laptop": {"price": 1200, "stock": 10},
    "Mouse": {"price": 25, "stock": 50},
    "Keyboard": {"price": 80, "stock": 20}
}

inventory["Monitor"]= {
    "price": 300,
    "stock": 15
}

inventory["Laptop"]["stock"] = 7

highest_inventory_value = 0
for individual,value in inventory.items():
    inventory[individual]["value"] = inventory[individual]["price"]*inventory[individual]["stock"] 
    if highest_inventory_value < inventory[individual]["value"]:
        highest_inventory_value = inventory[individual]["value"]

print(inventory)
print(highest_inventory_value)
```
